# Calculadora de IMC e Perfil de Usuário

Este notebook contém um sistema interativo em Python para coletar informações de usuários, classificar altura e calcular o Índice de Massa Corporal (IMC).

## 🚀 Funcionalidades

1.  **Entrada de Dados**: Coleta de nome, idade, peso e altura.
2.  **Lógica de Negócio**:
    *   Verificação de maioridade (18 anos).
    *   Classificação de estatura (Baixo, Mediano, Alto).
    *   Cálculo de IMC com classificação baseada nos padrões da OMS.
3.  **Interface Gráfica (GUI)**: Uma interface moderna e responsiva utilizando `ipywidgets` com tema escuro (Dark Mode).

## 🛠️ Tecnologias Utilizadas

*   **Python 3**
*   **ipywidgets**: Para a criação dos componentes de interface.
*   **IPython Display**: Para renderização de HTML/CSS personalizado.

## 📋 Como Usar

### Versão em Console
Execute a primeira célula do notebook para interagir via prompts de texto padrão (`input`).

### Versão com Interface Moderna
1. Execute as células da seção **Interface Gráfica Moderna**.
2. Preencha seu nome, selecione a idade no slider e insira seu peso/altura.
3. Clique no botão **Calcular Resultados** para visualizar o relatório estilizado.

---
*Desenvolvido para fins educacionais em ambiente Google Colab.*

```python
nome = input("Qual seu nome?")
print ("Olá", nome)
idade = input("Qual sua idade?")
print ("Você tem", idade, "anos")
idade_int = int(idade)
if idade_int >= 18:
    print ("Você é maior de idade")
    print ("Você pode votar")
else:
    print ("Você é menor de idade")
    print ("Você não pode votar")


peso = input("Qual seu peso?")
print ("Você pesa", peso, "kg")
peso_float = float(peso)

altura = input("Qual sua altura?")
print ("Você mede", altura, "m")
altura_float = float(altura.replace(',', '.'))

# Nova classificação de altura
if altura_float <= 1.65:
    print ("Você é baixo")
elif 1.65 < altura_float <= 1.79:
    print ("Você é mediano")
else:
    print ("Você é alto")

# Cálculo e Classificação do IMC
if altura_float > 0: # Evitar divisão por zero
    imc = peso_float / (altura_float ** 2)
    print (f"Seu IMC é: {imc:.2f}")

    if imc < 18.5:
        print ("Classificação IMC: Abaixo do peso")
    elif imc < 24.9:
        print ("Classificação IMC: Peso normal")
    elif imc < 29.9:
        print ("Classificação IMC: Sobrepeso")
    elif imc < 34.9:
        print ("Classificação IMC: Obesidade Grau I")
    elif imc < 39.9:
        print ("Classificação IMC: Obesidade Grau II")
    else:
        print ("Classificação IMC: Obesidade Grau III (Mórbida)")
else:
    print ("Não foi possível calcular o IMC: Altura inválida.")
```

### Interface Gráfica Moderna (Dark Theme)
Este bloco de código cria uma interface interativa utilizando `ipywidgets` e CSS personalizado para uma experiência de usuário superior.

```python
import ipywidgets as widgets
from IPython.display import display, HTML

# Estilização CSS Dark
style = """
<style>
    .main-container {
        background-color: #1e1e1e;
        padding: 20px;
        border-radius: 15px;
        border: 1px solid #333;
        color: #ffffff;
        font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;
    }
    .widget-label { color: #ffffff !important; font-weight: bold; }
    .custom-button {
        background-color: #007bff !important;
        color: white !important;
        font-weight: bold !important;
        border-radius: 8px !important;
    }
    .output-area {
        margin-top: 15px;
        padding: 10px;
        background-color: #2d2d2d;
        border-left: 5px solid #007bff;
        border-radius: 4px;
    }
</style>
"""
display(HTML(style))

# Widgets de Entrada
nome_w = widgets.Text(description='Nome:', placeholder='Digite seu nome')
idade_w = widgets.IntSlider(description='Idade:', min=0, max=120, value=25)
peso_w = widgets.FloatText(description='Peso (kg):', value=70.0)
altura_w = widgets.FloatText(description='Altura (m):', value=1.75)

btn_calcular = widgets.Button(description='Calcular Resultados', button_style='primary')
btn_calcular.add_class("custom-button")

out = widgets.Output()

def processar_dados(b):
    with out:
        out.clear_output()
        nome = nome_w.value
        idade = idade_w.value
        peso = peso_w.value
        altura = altura_w.value

        print(f"Olá, {nome}!")
        print(f"Status: {'Maior de idade' if idade >= 18 else 'Menor de idade'}")

        # Classificação de Altura
        if altura <= 1.65: cat_altura = "Baixo(a)"
        elif altura <= 1.79: cat_altura = "Mediano(a)"
        else: cat_altura = "Alto(a)"

        # Cálculo IMC
        if altura > 0:
            imc = peso / (altura ** 2)
            if imc < 18.5: classe = "Abaixo do peso"
            elif imc < 24.9: classe = "Peso normal"
            elif imc < 29.9: classe = "Sobrepeso"
            else: classe = "Obesidade"

            display(HTML(f"""
                <div class='output-area'>
                    <b>Altura:</b> {cat_altura}<br>
                    <b>IMC:</b> {imc:.2f} ({classe})
                </div>
            """))
        else:
            print("Erro: Altura inválida.")

btn_calcular.on_click(processar_dados)

# Layout
form = widgets.VBox([nome_w, idade_w, peso_w, altura_w, btn_calcular, out])
form.add_class("main-container")
display(form)
```
